```python
#Q6
hp_col = 'horsepower'

med_before = df[hp_col].median()
mode_val = df[hp_col].mode().iloc[0]

# Work only on the column, not the full DataFrame
med_after = df[hp_col].fillna(mode_val).median()

med_before, mode_val, med_after, (
    "Yes, it increased" if med_after > med_before 
    else "Yes, it decreased" if med_after < med_before 
    else "No"
)
```

```python
#Q7
df_asia = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)
X = df_asia.values
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv.dot(X.T).dot(y)
w.sum()
```
